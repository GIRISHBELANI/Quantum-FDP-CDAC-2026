# Linear Algebra

## Hilbert Space:

It is a complex multidimensional vector space where the inner product of any pair of elements is defined as the extrapolation of a 2D Space to n-Dimensions and complex coefficients.

- $|0\rangle, |1\rangle$ are considered as the standard basis states also as **orthonormal basis states**.

- Consider a 2D plane with an x-axis and y-axis, and we have a unit vector that aligns with the x-axis and a unit vector that aligns with the y-axis.

- Now we can represent any point on a 2D plane as the linear combination of the unit vectors of x and y.

- Hilbert space is similar to it. Here, the coeﬀicients are complex numbers, so the mathematics remains very similar.

- It is just like we are having different kinds of space, and, on a standard basis, roughly equivalent to our i & j vectors on a 2D plane, which represent orthonormal basis states. (They are perpendicular to each other).


## Bra-Ket Notation:

These are also called **Dirac** notations. Consider a vector $A$ in 3D Euclidean space, $A\in\mathbb{R}^3$. 

It is easy to see equivalences between ordinary notation and bra-ket notation in vector $A$. 

- Vector $A$ is the linear combination of the basis vectors that represent the coordinates.

![vector_space.jpg](vector_space.jpg)

$$\begin{aligned}
\overline{A}&=a_1\hat{i}+a_2\hat{j}+a_3\hat{k}\\
&=a_1\begin{pmatrix}1\\0\\0\end{pmatrix}+a_2\begin{pmatrix}0\\1\\0\end{pmatrix}+a_3\begin{pmatrix}0\\0\\1\end{pmatrix}\\
&=\begin{pmatrix}a_1\\0\\0\end{pmatrix}+\begin{pmatrix}0\\a_2\\0\end{pmatrix}+\begin{pmatrix}0\\0\\a_3\end{pmatrix}\ =\begin{pmatrix}a_1\\ a_2\\ a_3\end{pmatrix}\\
\end{aligned}$$

$$\Large{\boxed{\langle\text{bra}|\text{ket}\rangle}}$$

Now consider a vector B in an N-dimensional complex vector space $B\in \mathbb{C}^N$. Vector $B$ can be represented by a linear combination of basis vectors or a column matrix as: $$B = \sum_{n=1}^N b_n e_n\ =\begin{pmatrix}b_1\\b_2\\ \vdots\\ b_N\end{pmatrix}$$

### Ket Notation:

Conventional vector is written in bold type ($\bf{A}$) or with over/under-arrows $(\overrightarrow{A})\text{or}(\underline{A})$. In Dirac's notation for a vector uses vertical bars and angular brackets; a vector can be written as $|A\rangle$. Therefore,
$$|A\rangle = a_1|i\rangle + a_2|j\rangle + a_3|k\rangle = \begin{pmatrix}a_1\\a_2\\a_3\end{pmatrix}$$

Or simply one can write 3D vector in terms of ket in simple notation $$|A\rangle = a_1|1\rangle + a_2|2\rangle + a_3|3\rangle = \begin{pmatrix}a_1\\a_2\\a_3\end{pmatrix}$$

Similarly, for N dimensional vector Complex space $B\in\mathbb{C}^N$ in **ket** notation $$|B\rangle = \sum_{n=1}^{N} b_n|e_n\rangle =\begin{pmatrix}b_1\\b_2\\ \vdots \\b_N\end{pmatrix}$$

**Ket-0 representation:** $|0\rangle = \begin{bmatrix}1\\0\end{bmatrix}$

**Ket-1 representation:** $|1\rangle = \begin{bmatrix}0\\1\end{bmatrix}$

### Bra Notation:

It is the opposite operation of Ket $(|\rangle)$. It is actually an operation or a function that can be applied on Ket. For a given Ket we can compute **Bra** as the **conjugate Transpose of Ket**.

- **Bra-0 representation:** $\langle 0| = [ Conjugate \ of \begin{bmatrix} 1\\0 \end{bmatrix}]^T \implies \langle 0| = \begin{bmatrix} 1\\0 \end{bmatrix}^\dagger \implies \langle 0| = \begin{bmatrix} 1&0 \end{bmatrix}$

- **Bra-1 representation:** $\langle 1| = [ Conjugate \ of \begin{bmatrix} 0\\1 \end{bmatrix}]^T \implies \langle 1| = \begin{bmatrix} 0\\1 \end{bmatrix}^\dagger \implies \langle 1| = \begin{bmatrix} 0 & 1 \end{bmatrix} $

If $A$ is any ket $|A\rangle$ then its bra is given by $\langle 𝐴|$ which is
$$\Large{\boxed{\langle A| = A^\dagger = [\overline{A}]^T}}$$ and complex conjugate means if $\alpha = a+ib$ then, $\alpha^* \ or \ \alpha^\dagger = a-ib$.

#### **At N dimensional scale :** 

Now let us consider vectors with $n$ dimensions.

- Consider a general **ket-Si** which is actually a $n\times 1$ matrix.$$|\psi\rangle = \begin{bmatrix} a_1\\a_2\\ \vdots \\ a_n \end{bmatrix}$$

- Similarly a general **bra-Phi** which is actually a $1\times n$ matrix.$$\langle\phi| = \begin{bmatrix} b_1^* & b_2^* & \ldots & b_n^* \end{bmatrix} $$

## Why the conversion of Ket-to-bra or bra-to-ket is required for a vector?

In general, the vectors are represented in Ket format only i.e., as column matrix.
But if we want to perform any operation on that vector i.e, KET we cannot directly multiply another column matrix to a column matrix.
So the corresponding operation is converted into row matrix from column matrix. When we mulitply row matrix with column matrix we get the result. 

In quantum mechanics, the state of a system is represented by a ket vector, usually denoted as $|\psi\rangle$. These ket vectors are column matrices.

A bra vector, denoted as $\langle\psi|$, is the complex conjugate transpose of the corresponding ket vector. In other words, if $|\psi\rangle$ is a column vector, then $\langle\psi|$ is a row vector obtained by taking the conjugate transpose of $|\psi\rangle$.

When we perform operations involving kets, such as inner products or transformations, it's often necessary to use the complex conjugate of one of the vectors involved in the operation. This is where the bra-ket conversion comes in.

For example, to calculate the inner product of two vectors $|\psi\rangle$ and $|\phi\rangle$, we typically write it as $\langle\psi|\phi\rangle$. To do this, we need to convert one of the kets into a bra, so the operation can be performed properly.

Let's say we have:

$$
\vert \psi \rangle = \begin{pmatrix}
\psi_1 \\
\psi_2 \\
\vdots \\
\psi_n
\end{pmatrix}
$$


Then, the corresponding bra vector is:

$ \langle \psi \vert = (\psi_1^*, \psi_2^*, \ldots, \psi_n^*) $


where $\psi_i^*$ represents the complex conjugate of $\psi_n$.

Similarly, if we have a ket vector $|\phi\rangle$:

$$ \vert \phi \rangle = \begin{pmatrix}
\phi_1 \\
\phi_2 \\
\vdots \\
\phi_n
\end{pmatrix} $$


Then, the corresponding bra vector is:
$ \langle \phi \vert = (\phi_1^*, \phi_2^*, \ldots, \phi_n^*) $

Once we have the bra vector, we can perform operations such as inner products by taking the dot product of the bra and ket vectors, or transformations by multiplying the bra and ket matrices.

So, the conversion between ket and bra notation, along with complex conjugate matrices, enables us to perform various operations in quantum mechanics.

# Properties of Quantum Gates

Quantum gates are the fundamental building blocks of quantum circuits. Unlike classical logic gates, quantum gates operate on qubits and must obey the laws of quantum mechanics.

## 1. Reversible Operations

- Every quantum gate is **reversible**, meaning the input state can always be recovered from the output state.

$$
|\psi_{\text{out}}\rangle = U|\psi_{\text{in}}\rangle
$$

- Since every quantum gate is represented by a unitary matrix $U$, an inverse gate always exists:

$$
U^{-1}U = UU^{-1} = I
$$

- For example, the Pauli-X gate is its own inverse:

$$
X^{-1}=X
$$

$$
XX=I
$$


## 2. Unitary Nature

- All quantum gates are represented by **unitary matrices**. A matrix $U$ is unitary if $$U^\dagger U = UU^\dagger = I$$ where $U^\dagger$ denotes the conjugate transpose of $U$.

- This property ensures that quantum operations preserve the total probability of a quantum state.


## 3. Preserve Normalization

- If a quantum state satisfies $$|\alpha|^2+|\beta|^2=1$$ before applying a gate, it will remain normalized after the operation.

$$
\langle\psi|\psi\rangle = 1
$$

- This guarantees that measurement probabilities always sum to one.


## 4. Operate on Superposition States

- Quantum gates can act on superposition states without destroying them. For example, $$H|0\rangle=\frac{|0\rangle+|1\rangle}{\sqrt2}$$

- Quantum gates transform all amplitudes simultaneously, enabling quantum parallelism.


## 5. Can Change Relative Phase

- Unlike classical gates, quantum gates can modify the phase of a quantum state. For example, the Z-gate performs $$Z|0\rangle = |0\rangle$$ $$Z|1\rangle = -|1\rangle$$

- The probabilities remain unchanged, but the phase information changes.


## 6. Can Create and Manipulate Entanglement

Multi-qubit gates such as CNOT can generate entangled states.

For example,

$$
H|0\rangle=\frac{|0\rangle+|1\rangle}{\sqrt2}
$$

Applying a CNOT gate gives

$$
\frac{|00\rangle+|11\rangle}{\sqrt2}
$$

which is an entangled Bell state.

## 7. Linear Transformations

Quantum gates are linear operators.

For any states $|\psi\rangle$ and $|\phi\rangle$,

$$
U(a|\psi\rangle+b|\phi\rangle) = aU|\psi\rangle+bU|\phi\rangle
$$

where $a$ and $b$ are complex numbers.

## 8. No Information Loss

Since quantum gates are reversible and unitary, information is never destroyed during computation.

This is fundamentally different from classical irreversible gates such as AND, OR, and NAND.

# Single-Qubit Quantum Gates:

## 1. Pauli-X Gate (X Gate)

The Pauli-X gate acts as a **Quantum NOT Gate**. It flips the state of a qubit:

$$
|0\rangle \leftrightarrow |1\rangle
$$

### Matrix Representation

$$
X=
\begin{bmatrix}
0 & 1\\
1 & 0
\end{bmatrix}
$$



### Truth Table

| Input State | Output State |
|------------|------------|
| $|0\rangle$ | $|1\rangle$ |
| $|1\rangle$ | $|0\rangle$ |

### Examples

Applying X on $|0\rangle$

$$
X|0\rangle=
\begin{bmatrix}
0 & 1\\
1 & 0
\end{bmatrix}
\begin{bmatrix}
1\\
0
\end{bmatrix}
=\begin{bmatrix}
0\\
1
\end{bmatrix}
=|1\rangle
$$

Applying X on $|1\rangle$

$$
X|1\rangle=
\begin{bmatrix}
0 & 1\\
1 & 0
\end{bmatrix}
\begin{bmatrix}
0\\
1
\end{bmatrix}
=\begin{bmatrix}
1\\
0
\end{bmatrix}
=|0\rangle
$$

## 2. Pauli-Y Gate (Y Gate)

The Pauli-Y gate performs a bit flip together with a phase change.

### Matrix Representation

$$
Y=
\begin{bmatrix}
0 & -i\\
i & 0
\end{bmatrix}
$$


### Truth Table

| Input State | Output State |
|------------|------------|
| $|0\rangle$ | $i|1\rangle$ |
| $|1\rangle$ | $-i|0\rangle$ |

### Examples

$$
Y|0\rangle=
\begin{bmatrix}
0 & -i\\
i & 0
\end{bmatrix}
\begin{bmatrix}
1\\
0
\end{bmatrix}
=i|1\rangle
$$

$$
Y|1\rangle=
\begin{bmatrix}
0 & -i\\
i & 0
\end{bmatrix}
\begin{bmatrix}
0\\
1
\end{bmatrix}
=-i|0\rangle
$$

## 3. Pauli-Z Gate (Z Gate)

The Pauli-Z gate changes the phase of the $|1\rangle$ state while leaving $|0\rangle$ unchanged.

### Matrix Representation

$$
Z=
\begin{bmatrix}
1 & 0\\
0 & -1
\end{bmatrix}
$$

### Truth Table

| Input State | Output State |
|------------|------------|
| $|0\rangle$ | $|0\rangle$ |
| $|1\rangle$ | $-|1\rangle$ |

### Examples

$$
Z|0\rangle=
\begin{bmatrix}
1\\
0
\end{bmatrix}
=|0\rangle
$$

$$
Z|1\rangle=
\begin{bmatrix}
0\\
-1
\end{bmatrix}
=-|1\rangle
$$

## 4. Hadamard Gate (H Gate)

The Hadamard gate creates an equal superposition of basis states.

### Matrix Representation

$$
H=
\frac{1}{\sqrt2}
\begin{bmatrix}
1 & 1\\
1 & -1
\end{bmatrix}
$$

### Truth Table

| Input State | Output State |
|------------|------------|
| $|0\rangle$ | $\frac{|0\rangle+|1\rangle}{\sqrt2}$ |
| $|1\rangle$ | $\frac{|0\rangle-|1\rangle}{\sqrt2}$ |

### Examples

$$
H|0\rangle
=\frac{1}{\sqrt2}
\begin{bmatrix}
1\\
1
\end{bmatrix}
=\frac{|0\rangle+|1\rangle}{\sqrt2}
$$

$$
H|1\rangle
=\frac{1}{\sqrt2}
\begin{bmatrix}
1\\
-1
\end{bmatrix}
=\frac{|0\rangle-|1\rangle}{\sqrt2}
$$

## 5. Phase Gate (S Gate):

The S gate adds a phase of $90^\circ$ ($\pi/2$ radians) to the state $|1\rangle$.

### Matrix Representation

$$
S=\begin{bmatrix}1 & 0\\0 & i\end{bmatrix}
$$

### Truth Table

| Input State | Output State |
|------------|------------|
| $|0\rangle$ | $|0\rangle$ |
| $|1\rangle$ | $i|1\rangle$ |

### Examples

$$
S|0\rangle=|0\rangle
$$

$$
S|1\rangle=i|1\rangle
$$

## 6. T Gate ($\pi/8$ Gate)

The T gate adds a phase of

$$
e^{i\pi/4}
$$

to the state $|1\rangle$.

### Matrix Representation

$$
T=
\begin{bmatrix}
1 & 0\\
0 & e^{i\pi/4}
\end{bmatrix}
$$

### Truth Table

| Input State | Output State |
|------------|------------|
| $|0\rangle$ | $|0\rangle$ |
| $|1\rangle$ | $e^{i\pi/4}|1\rangle$ |

### Examples

$$
T|0\rangle=|0\rangle
$$

$$
T|1\rangle=e^{i\pi/4}|1\rangle
$$

# Qubit Representation on Bloch Sphere:

The qubit state **$|\psi\rangle$** can be visualized on the surface of a unit sphere. The Bloch sphere representation is based on the following mapping:

- The poles of the sphere **$|0\rangle$** and **$|1\rangle$** represent the basis states of the qubit.
- Any state on the surface of the sphere represents a valid qubit state.

## Bloch Sphere Coordinates:

The qubit state vector **$|\psi\rangle$** is associated with three spherical coordinates: **$\theta$, $\phi$**, and the radius *r = 1*.

1. **Radius (r = 1):** The Bloch sphere has a fixed radius of 1, representing the normalization constraint of the qubit state vector.

2. **Angles ($\theta$ and $\phi$):** 
   - $\theta$ represents the angle between the state vector and the *z*-axis (polar angle).
   - $\phi$ represents the angle in the **xy**-plane measured counterclockwise from the positive *x*-axis (azimuthal angle).

The **general equation of a single qubit on the Bloch Sphere** is represented as

$$
\boxed{|\psi\rangle=\cos\frac{\theta}{2}|0\rangle+e^{i\phi}\sin\frac{\theta}{2}|1\rangle}
$$

### States on the Bloch Sphere:

- **Basis States:** The poles of the Bloch sphere represent the **$|0\rangle$** and **$|1\rangle$** states.
- **Superposition States:** States between the poles represent superpositions of **$|0\rangle$** and **$|1\rangle$** states.

### Manipulation of Qubit States:

- **Quantum Gates:** Operations or gates applied to qubits correspond to transformations on the Bloch sphere. For instance, a Hadamard gate rotates the state vector around specific axes on the sphere.

### Importance:

- **Visualization:** The Bloch sphere provides an intuitive visualization of qubit states, aiding in understanding quantum operations and quantum algorithms.
- **Analysis:** It's useful for analyzing quantum gates' effects on qubit states and for representing various quantum phenomena.

The Bloch sphere representation serves as a valuable tool in understanding and visualizing qubit states and their manipulation in quantum computing.

## Demonstration of $X\cdot|0\rangle$ state:

````python
from qiskit.quantum_info import Operator, Statevector
from qutip import Bloch, Qobj
from IPython.display import display, Math
from qiskit.visualization import plot_bloch_multivector

# Define |0⟩ state
psi = Statevector([1, 0])
display(Math(r"\text{Initial State: }"+psi.draw('latex_source')))
display(plot_bloch_multivector(psi))

# Define X gate operator
X = Operator([[0, 1],[1, 0]])
# Print operator
display(Math(r"\text{X-Gate: }"+X.draw('latex_source')))
#display(plot_bloch_multivector(X))

# Apply X gate
psi_final = psi.evolve(X)
display(plot_bloch_multivector(psi_final))

# Print LaTeX representation
display(Math(r"\text{Final State: }"+psi_final.draw('latex_source')))
````

### What is exactly happening here?

$$|0\rangle = \begin{bmatrix}1 \\ 0\end{bmatrix}$$

$$X = \begin{bmatrix}0 & 1 \\ 1 & 0\end{bmatrix}$$

The **evolve** function performs below operation:

$$\boxed{\begin{aligned}
X\cdot|0\rangle &= \begin{bmatrix}0 & 1 \\ 1 & 0\end{bmatrix} \times \begin{bmatrix}1 \\ 0\end{bmatrix}\\
&=\begin{bmatrix}0 \\ 1\end{bmatrix} = |1\rangle
\end{aligned} }$$

## Sequence of Execution of Quantum Gates:

````python
import sympy

v = Statevector([1, 0])
print("###################### Initial State: #####################")
display(v.draw("latex"))
display(plot_bloch_multivector(psi))

Y = Operator([[0, -1.0j], [1.0j, 0]])
H = Operator([[1 / np.sqrt(2), 1 / np.sqrt(2)], [1 / np.sqrt(2), -1 / np.sqrt(2)]])
S = Operator([[1, 0], [0, 1.0j]])
T = Operator([[1, 0], [0, (1 + 1.0j) / np.sqrt(2)]])

print("############## Applying H-Gate on initial state ############")
v1 = v.evolve(H)
display(v1.draw("latex"))
display(plot_bloch_multivector(v1))
print("############## v2 State -> Applying T-Gate on v1 state #################")
v2 = v1.evolve(T)
display(v2.draw("latex"))
display(plot_bloch_multivector(v2))
print("############## v3 State -> Applying H-Gate on v2 state #################")
v3 = v2.evolve(H)
display(v3.draw("latex"))
display(plot_bloch_multivector(v3))
print("############### v4 -> Applying s-Gate on v3 state ######################")
v4 = v3.evolve(S)
display(v4.draw("latex"))
display(plot_bloch_multivector(v4))
print("############### v5 -> Applying Y-Gate on v4 state to get final output state ########")
v5 = v4.evolve(Y)
display(v5.draw("latex"))
display(plot_bloch_multivector(v5))
````
