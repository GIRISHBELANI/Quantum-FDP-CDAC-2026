# Inner and Outer Products:

## Inner Product:

This Operation applies **bra-phi** $\langle\Phi|$ to **Ket-si** $|\psi\rangle$ and this is an inner product of both the vectors or better can be interpreted as the **Dot product of both vectors.

$\langle A|B\rangle$ = inner product of $|A\rangle$ with $|B\rangle$
$$\boxed{\langle A|B\rangle = \sum_{n=1}^{N} {a_n}^*{e_n}^* {a_n}{e_n} = \begin{pmatrix}{a_1}^* & {a_2}^* & \ldots &{a_N}^* \end{pmatrix} \begin{pmatrix}b_1 \\ b_2\\ \vdots \\ b_N \end{pmatrix} = {a_1}^* b_1 + {a_2}^* b_2 + \ldots + {a_N}^* b_N}$$

```text
The inner product always gives a scalar value as output.
```

$$\implies {a_n}^* \rightarrow conjugate(a_i)$$
$$\implies \langle A| \rightarrow conjugate(|A\rangle) = bra-A$$

**Example:**

$$|0\rangle=\begin{pmatrix}1 \\0 \end{pmatrix} \implies \langle 0|=\begin{pmatrix}1 & 0\end{pmatrix}$$
$$\therefore \langle 0|0\rangle = \begin{pmatrix}1 & 0\end{pmatrix}\begin{pmatrix}1 \\0 \end{pmatrix} = 1$$

Similarly,
$$|1\rangle=\begin{pmatrix}0 \\1 \end{pmatrix} \implies \langle 1|=\begin{pmatrix}0 & 1\end{pmatrix}$$
$$\therefore \langle 1|0\rangle = \begin{pmatrix}0 & 1\end{pmatrix}\begin{pmatrix}1 \\0 \end{pmatrix} = 0$$

- If the inner product of any two unequal vectors (i.e bra-ket operation of two vectors) is **0**, then the vectors are Orthogonal to each other.

- The inner product of any two same vectors is always **1**, which is the indication for normalization. All quantum states are always normalized.

## Outer Product:

$|A\rangle\langle B|$ = Outer-product of **A and B**.

They are extremely useful in describing **density operators**, **quantum-gates**, etc.

```text
The outer product will always give a square matrix as the result.
```

**Example:**

A not gate is expressed as $$\mathsf{NOT}=|0\rangle\langle 1| + |1\rangle\langle 0|$$

$$|0\rangle =\begin{pmatrix}1 \\ 0 \end{pmatrix} \implies \langle 0| =\begin{pmatrix}1 & 0\end{pmatrix}$$

Similarly,

$$|1\rangle =\begin{pmatrix}0 \\ 1 \end{pmatrix} \implies \langle 1| =\begin{pmatrix}0 & 1\end{pmatrix}$$

$$\therefore |0\rangle\langle 1| = \begin{pmatrix}1 \\ 0 \end{pmatrix}\begin{pmatrix}0 & 1\end{pmatrix} = \begin{pmatrix}0 & 1\\0 & 0\end{pmatrix}$$

$$\therefore |1\rangle\langle 0| = \begin{pmatrix}0 \\ 1 \end{pmatrix}\begin{pmatrix}1 & 0\end{pmatrix} = \begin{pmatrix}0 & 0\\1 & 0\end{pmatrix}.$$

Thus,

$$\boxed{\mathsf{NOT}=|0\rangle\langle 1| + |1\rangle\langle 0| = \begin{pmatrix}0 & 1\\0 & 0\end{pmatrix} + \begin{pmatrix}0 & 0\\1 & 0\end{pmatrix} = \begin{pmatrix}0 & 1\\1 & 0\end{pmatrix}}$$

$$ \boxed{\color{red}{Ket-Bra \ \vert\Psi\rangle\langle\Phi\lvert \ = \ \begin{bmatrix}a_1 \\ a_2\\ . \\ .\\ .\\ a_n \end{bmatrix}\ \begin{bmatrix}b_1^* & b_2* & . & . & . & b_n* \end{bmatrix}\ = \begin{bmatrix}a_1b_1^* & a_1b_2^* & . & . & . & a_1b_N* \\ a_2b_1^* & a_2b_2^* & . & . & . & a_2b_N* \\ . \\ . \\ . \\ a_Nb_1^* & a_Nb_2^* & . & . & . & a_Nb_N* \end{bmatrix} }}$$

# Tensor Product:

Mathematically tensor Product is also referred as Kronecker Product.

$$\begin{bmatrix}a\\b\end{bmatrix}\otimes\begin{bmatrix}c\\d\end{bmatrix} = \begin{bmatrix}a\begin{bmatrix}c\\d\end{bmatrix} \\ b \begin{bmatrix}c\\d\end{bmatrix} \end{bmatrix} = \begin{bmatrix}ac\\ad\\bc\\bd\end{bmatrix}$$

$$|01\rangle = |0\rangle\otimes|1\rangle = \begin{pmatrix}1\\0\end{pmatrix}\otimes\begin{pmatrix}0\\1\end{pmatrix} = \begin{pmatrix}0\\1\\0\\0\end{pmatrix}$$

$$|10\rangle = |1\rangle\otimes|0\rangle = \begin{pmatrix}0\\1\end{pmatrix}\otimes\begin{pmatrix}1\\0\end{pmatrix} = \begin{pmatrix}0\\0\\1\\0\end{pmatrix}$$

$$|00\rangle = |0\rangle\otimes|0\rangle = \begin{pmatrix}1\\0\end{pmatrix}\otimes\begin{pmatrix}1\\0\end{pmatrix} = \begin{pmatrix}1\\0\\0\\0\end{pmatrix}$$

Similarly,

$$|111\rangle = |1\rangle\otimes|1\rangle\otimes|1\rangle = \begin{pmatrix}0\\1\end{pmatrix}\begin{pmatrix}0\\1\end{pmatrix}\otimes\begin{pmatrix}0\\1\end{pmatrix} = \begin{pmatrix}0\\0\\0\\1\end{pmatrix}\otimes\begin{pmatrix}0\\1\end{pmatrix} = \begin{pmatrix}0\\0\\0\\0\\0\\0\\0\\1\end{pmatrix}$$

- Dimension of Hilbert space (a vector space where inner products are well defined, complex projection is allowed and there is no restriction on the dimension) increases quickly in the quantum world.

Let $|O_1O_2\ldots O_n\rangle$ is a column matrix with $2^n$ rows and any operator (gate or circuit) that can operate on this and transform it into a state vector of same dimension will be a $2^n\times 2^n$ matrix.

# Measurement of Quantum States:

## What is Quantum Measurement?

In quantum computing, a **measurement** is the process of obtaining a classical result from a quantum state. A qubit can exist in a **superposition** of the states $\lvert 0\rangle$ and $\lvert 1\rangle$:

$$
\lvert\psi\rangle=\alpha\lvert0\rangle+\beta\lvert1\rangle
$$ where $\alpha$ and $\beta$ are complex probability amplitudes satisfying

$$
|\alpha|^2+|\beta|^2=1
$$

- When the qubit is measured in the computational basis, the result can only be:

$$
0 \quad \text{or} \quad 1
$$

- The probabilities are $P(0)=|\alpha|^2$ and $P(1)=|\beta|^2$.

> **Important:** Measurement does not return the amplitudes $\alpha$ and $\beta$. It returns a classical outcome such as $0$ or $1$ according to its probabilities.


## Why is Measurement Needed?

A quantum computer performs calculations using **qubits**, but the final result must usually be used by a classical computer or a human. Therefore, we need a process to convert the quantum state into **classical information**.

- The basic flow is:

$$
\boxed{
\text{Quantum State}
\rightarrow
\text{Quantum Operations}
\rightarrow
\text{Measurement}
\rightarrow
\text{Classical Result}
}
$$

- Without measurement, we cannot directly extract the final computational result from a quantum circuit.


## Measurement of a Single Qubit

- Suppose a qubit is in the state

$$
\lvert\psi\rangle=\frac{1}{\sqrt{2}}\lvert0\rangle+\frac{1}{\sqrt{2}}\lvert1\rangle
$$

- The probabilities are $$P(0)=\left|\frac{1}{\sqrt{2}}\right|^2=\frac{1}{2}$$ and $$P(1)=\left|\frac{1}{\sqrt{2}}\right|^2=\frac{1}{2}$$

- Therefore, measuring the qubit gives: $$0 \quad \text{with probability }50%$$ and $$1 \quad \text{with probability }50%$$

- A single measurement gives only **one outcome**.


## Measurement Causes State Collapse

Before measurement, the qubit may be in a superposition:

$$
\lvert\psi\rangle=\alpha\lvert0\rangle+\beta\lvert1\rangle
$$

- After measuring and obtaining $0$, the state becomes

$$
\lvert\psi\rangle=\lvert0\rangle
$$

- Similarly, if the measurement gives $1$,

$$
\lvert\psi\rangle=\lvert1\rangle
$$

- This is commonly called **collapse of the quantum state**.

> **Intuition:** Before measurement, the qubit contains quantum probability amplitudes. Measurement produces one definite classical outcome.

### Example 1: Measuring a Qubit in State $\lvert0\rangle$

- Consider $$\lvert\psi\rangle=\lvert0\rangle$$

- The state contains no $\lvert1\rangle$ component. Therefore, $P(0)=1$ and $P(1)=0$.

- So every measurement gives $\boxed{0}$.

### Example 2: Measuring a Superposition

- Consider

$$
\lvert\psi\rangle=\frac{1}{\sqrt2}\lvert0\rangle+\frac{1}{\sqrt2}\lvert1\rangle
$$

- Then, $P(0)=\frac{1}{2}$ and $P(1)=\frac{1}{2}$

- Suppose we measure the qubit 10 times. A possible set of results could be:

```text
0  1  1  0  0  1  0  1  1  0
```

- The exact sequence is random, but after many measurements, the frequency of $0$ and $1$ approaches approximately $50%$ each.

### Example 3: Unequal Probabilities

- Consider $$\lvert\psi\rangle=\frac{\sqrt3}{2}\lvert0\rangle+\frac12\lvert1\rangle$$

- The probabilities are $P(0)=\left|\frac{\sqrt3}{2}\right|^2\frac34$ and $P(1)= \left|\frac12\right|^2\frac14$

- Therefore,

$$
\boxed{P(0)=75\% ,\qquad P(1)=25\%}
$$

- After many measurements, we would expect approximately $75%$ zeros and $25%$ ones.

###  Why Do We Run a Quantum Circuit Multiple Times?

Because measurement is **probabilistic**. A single execution produces only one outcome. Therefore, quantum circuits are usually executed many times, called **shots**.

- For example, if a circuit is expected to produce

    $$
    P(0)=0.7,\qquad P(1)=0.3
    $$

- And we run it 1000 times, we may obtain approximately:

    ```text
    0 → 700 times
    1 → 300 times
    ```

- The collection of measurement results gives an estimate of the underlying probability distribution.


# Single and Multi-Qubit Systems:

## Single Qubit Systems

A Single qubit is nothing but a vector in a complex 2D Hilbert space. A general Qubit state is represented as an **Arbitrary Ket**, which is a linear combination of orthonormal basis states with complex Coefficients.

$$\text{General Qubit Representation} \implies \boxed{ \begin{aligned} |\psi\rangle &= \alpha|0\rangle + \beta|1\rangle \\ &= \alpha\times\begin{bmatrix}1\\0\end{bmatrix} + \beta\times\begin{bmatrix}0\\1\end{bmatrix}\\ &= \begin{bmatrix}\alpha\\\beta\end{bmatrix}\end{aligned}}$$

Where:
- $\alpha$ and $\beta$ are complex numbers known as probability amplitudes, satisfying the normalization condition $$(|\alpha|^2 + |\beta|^2 = 1)$$

So if we have a state like: $$ \lvert \Psi \rangle \ = \ a \lvert 0 \rangle \ + \ b \lvert 1 \rangle \ $$ and we are making a measurement in the computational basis, then we are not going to get a linear combination as a result.

We are going to get the state $|0\rangle$ or the state $|1\rangle$, which is called the **collapse of the wave function**.

1. We will get state $|0\rangle$ with the probability given by $$\frac{a}{\sqrt{a^2+b^2}}$$ and this is the probability amplitude of this state being measured as $|0\rangle$.
   
2. Similarly state $|1\rangle$ has probability amplitude given by $$\frac{b}{\sqrt{a^2+b^2}}$$

## Multi-Qubit Systems

A Multi-Qubit System is formed by combining multiple single qubits using the **Tensor Product ($\otimes$)** operation. While a single qubit exists in a $2D$ Hilbert space, an $n$-qubit system exists in a $2^n$-dimensional Hilbert space.

For a two-qubit system, the computational basis states are:

$$
|00\rangle =\begin{bmatrix}1\\0\\0\\0\end{bmatrix},
\quad
|01\rangle =\begin{bmatrix}0\\1\\0\\0\end{bmatrix},
\quad
|10\rangle =\begin{bmatrix}0\\0\\1\\0\end{bmatrix},
\quad
|11\rangle =\begin{bmatrix}0\\0\\0\\1\end{bmatrix}
$$

The general representation of a two-qubit state is:

$$
\boxed{
\begin{aligned}
|\psi\rangle&=\alpha|00\rangle+\beta|01\rangle+\gamma|10\rangle+\delta|11\rangle
\\[4pt]
&=\begin{bmatrix}\alpha\\\beta\\\gamma\\\delta\end{bmatrix}
\end{aligned}
}
$$

where the probability amplitudes satisfy the normalization condition

$$
|\alpha|^2+|\beta|^2+|\gamma|^2+|\delta|^2 = 1
$$

### Example

Consider the two-qubit state

$$
|\psi\rangle=
\frac{1}{\sqrt2}|00\rangle
+
\frac{1}{\sqrt2}|11\rangle
$$

Its state vector representation is

$$
|\psi\rangle=\frac{1}{\sqrt2}\begin{bmatrix}1\\0\\0\\1\end{bmatrix}=\begin{bmatrix}0.7071\\0\\0\\0.7071\end{bmatrix}
$$

When measured in the computational basis:

- State $|00\rangle$ is obtained with probability

$$
\left|\frac{1}{\sqrt2}\right|^2=\frac12
$$

- State $|11\rangle$ is obtained with probability

$$
\left|\frac{1}{\sqrt2}\right|^2=\frac12
$$

- States $|01\rangle$ and $|10\rangle$ have probability

$$
0
$$

Thus, an $n$-qubit system requires a state vector of length $2^n$, allowing quantum computers to represent exponentially many basis states simultaneously.

The CNOT gate flips the target qubit only when the control qubit is $|1\rangle$.

### Matrix Representation

The CZ gate applies a phase flip only when both qubits are in state $|1\rangle$.

It is the quantum equivalent of a controlled-controlled NOT operation.


# Entanglement and Bell States:

## Entanglement:

> Quantum entanglement is a phenomenon that occurs when two or more quantum particles become deeply interconnected in such a way that the state of one particle is intrinsically linked to the state of another, regardless of the distance between them.

- This connection persists even if the particles are separated by vast distances, which famously led Einstein to refer to it as **"spooky action at a distance"**. Now, when it comes to qubits, entanglement plays a pivotal role in the power and potential of quantum computing. Here's how:

1. **Entangled Qubits**:

   - In a quantum system, when qubits become entangled, the state of one qubit becomes dependent on the state of the other(s), and vice versa. This connection exists irrespective of the physical separation between the qubits.

   - For instance, if two qubits become entangled, measuring the state of one qubit instantly determines the state of the other, regardless of the distance between them. This instantaneous correlation is a hallmark of entanglement.

2. **Effect on Qubits**:

   - Entanglement allows for the creation of highly correlated quantum states between qubits. This correlation persists even if the entangled qubits are separated by large distances.

   - When qubits are entangled, they share a joint quantum state, and measurements on one qubit will instantaneously affect the state of the other qubit, ensuring they remain correlated.

3. **Applications in Quantum Computing**:

   - Entanglement is crucial in quantum computing algorithms. It enables the manipulation of qubits in a way that exploits their interconnectedness to perform computations that classical computers find extremely challenging or impossible to achieve efficiently.

   - Quantum algorithms, such as Shor's algorithm for factoring large numbers or quantum teleportation, leverage entanglement to perform computations more efficiently than classical counterparts.

4. **Quantum Communication**:

   - Entanglement also has implications for quantum communication and cryptography. It allows for secure communication protocols, such as quantum key distribution (QKD), where entangled qubits are used to securely transmit encryption keys.

5. **Challenges**:

   - While entanglement is a powerful resource in quantum computing, it's delicate and easily disrupted by interactions with the environment, causing a phenomenon known as decoherence. Efforts in quantum computing research focus on methods to preserve and harness entanglement for practical computational purposes.

In essence, entanglement is a remarkable feature of quantum physics that underlies the potential of quantum computing. It enables the creation of highly correlated quantum states between qubits, offering computational advantages and opening up possibilities for secure communication and cryptography in the quantum realm.

### Entangled States:

An **entangled state** is a quantum state of two or more qubits in which the state of one qubit cannot be described independently of the other qubits.

- For two qubits, a general state can be written as

$$
|\psi\rangle
=\alpha|00\rangle
+\beta|01\rangle
+\gamma|10\rangle
+\delta|11\rangle
$$

- A two-qubit state is **separable** if it can be written as a product of two single-qubit states:

$$
|\psi\rangle=|\psi_1\rangle\otimes|\psi_2\rangle
$$

- If it **cannot be separated into a product state**, then it is an **entangled state**.

$$
\boxed{
|\psi\rangle\neq|\psi_1\rangle\otimes|\psi_2\rangle
}
$$

> **Key idea:** An entangled state describes the two qubits as a single quantum system rather than as two independent qubits.

- Consider $|\psi\rangle=\frac{1}{\sqrt{2}}\left(|00\rangle+|11\rangle\right)$

- Suppose we try to write it as

$$
|\psi\rangle
=(a|0\rangle+b|1\rangle)
\otimes
(c|0\rangle+d|1\rangle)
$$

- Expanding,

$$
|\psi\rangle
=ac|00\rangle
+ad|01\rangle
+bc|10\rangle
+bd|11\rangle
$$

- To obtain $\frac{1}{\sqrt2}|00\rangle+\frac{1}{\sqrt2}|11\rangle,$ we would require $$
ac=\frac{1}{\sqrt2},
\qquad
ad=0,
\qquad
bc=0,
\qquad
bd=\frac{1}{\sqrt2}.
$$

- These conditions cannot be satisfied simultaneously. Therefore, $$\boxed{\frac{1}{\sqrt2}(|00\rangle+|11\rangle)}$$ is an **entangled state**.

### Bell States

The **Bell states** are four specific maximally entangled two-qubit states. They are:

#### Bell State 1

$$
|\Phi^+\rangle
=\frac{1}{\sqrt2}
(|00\rangle+|11\rangle)
$$

#### Bell State 2

$$
|\Phi^-\rangle
=\frac{1}{\sqrt2}
(|00\rangle-|11\rangle)
$$

#### Bell State 3

$$
|\Psi^+\rangle
=\frac{1}{\sqrt2}
(|01\rangle+|10\rangle)
$$

#### Bell State 4

$$
|\Psi^-\rangle
=\frac{1}{\sqrt2}
(|01\rangle-|10\rangle)
$$

These four states form the **Bell basis** for a two-qubit system.

### Example: Creating a Bell State

- The Bell state $$|\Phi^+\rangle=\frac{1}{\sqrt2}(|00\rangle+|11\rangle)$$ can be created using a simple quantum circuit.

- Start with $|00\rangle$. Apply a Hadamard gate to the first qubit:

$$
|00\rangle
\xrightarrow{H\otimes I}
\frac{1}{\sqrt2}
(|00\rangle+|10\rangle)
$$

- Then apply a CNOT gate:

$$
\frac{1}{\sqrt2}
(|00\rangle+|10\rangle)
\xrightarrow{CNOT}
\frac{1}{\sqrt2}
(|00\rangle+|11\rangle)
$$

- Thus,

$$
\boxed{
|00\rangle
\xrightarrow{H}
\xrightarrow{CNOT}
|\Phi^+\rangle
}
$$

#### Measurement of an Entangled State

- Consider

$$
|\Phi^+\rangle
=\frac{1}{\sqrt2}
(|00\rangle+|11\rangle)
$$

- When measured, the possible outcomes are $00$ or $11$ with probabilities $P(00)=\frac12$ and $P(11)=\frac12.$ Therefore, if the first qubit is measured as $0$, the second qubit will also be $0$. If the first qubit is measured as $1$, the second qubit will also be $1$.

> **Important:** This does not mean that the individual qubits had definite classical values before measurement. The pair was described by a joint quantum state.

# Statevector Simulator vs Aer Simulator

## Statevector Simulator

- A **Statevector Simulator** is a classical computer program that simulates the mathematical state of a quantum system. It represents the quantum state as a vector of complex-number amplitudes and updates that vector whenever quantum gates are applied.

- For $n$ qubits, the state is represented by a vector of size $2^n$.

- For example, for two qubits, $$|\psi\rangle=\alpha|00\rangle+\beta|01\rangle+\gamma|10\rangle+\delta|11\rangle$$ is represented by the statevector

$$
\begin{bmatrix}
\alpha\\
\beta\\
\gamma\\
\delta
\end{bmatrix}
$$


> - Gives access to the **complete quantum state**.
> - Useful for examining amplitudes and probabilities.
> - Does not represent the behavior of a real noisy quantum device.
> - The memory requirement grows exponentially with the number of qubits.

### Example

- For the state $$|\psi\rangle=\frac{1}{\sqrt2}(|0\rangle+|1\rangle)$$ the statevector is

$$
\begin{bmatrix}
\frac{1}{\sqrt2}\\
\frac{1}{\sqrt2}
\end{bmatrix}
$$

> A Statevector Simulator allows us to directly inspect these amplitudes.



````python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create a one-qubit circuit
qc = QuantumCircuit(1)

# Apply a Hadamard gate
qc.h(0)

# Simulate the quantum state
state = Statevector.from_instruction(qc)

print(qc.draw())
print("Statevector:", state.data)
print("Probabilities:", state.probabilities())
````



## Aer Simulator

- **Qiskit Aer** is a high-performance simulator framework used to simulate quantum circuits.

- Unlike a simple statevector calculation, Aer provides different simulation methods and can model effects such as **measurement and noise**, depending on the chosen simulator configuration.

> - Used to simulate complete quantum circuits.
> - Supports different simulation methods such as statevector, density matrix, and stabilizer methods.
> - Can execute circuits multiple times using **shots**.
> - Can include realistic noise models.
> - Useful for testing and debugging quantum algorithms before running them on real quantum hardware.

### Example

- Suppose a circuit creates

$$
|\psi\rangle=
\frac{1}{\sqrt2}(|0\rangle+|1\rangle)
$$

- Running it with Aer for **1000** shots might produce approximately:

$$
0 \rightarrow 503 \quad
1 \rightarrow 497
$$

- The exact counts vary because measurement is probabilistic.


Create a circuit that applies a Hadamard gate.

````python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1)
qc.h(0)

simulator = AerSimulator(method="statevector")

result = simulator.run(qc).result()   # submits the circuit for simulation

print(result)
````

To explicitly retrieve the final statevector, save it in the circuit:

**Simulating quantum noise**

One important feature of Aer is its ability to simulate imperfect gates and measurements.

For example, a bit-flip error changes $|0\rangle$ into $|1\rangle$, or vice versa, with a specified probability.

````python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, pauli_error

# Define a bit-flip error with probability 0.1
bit_flip = pauli_error([("X", 0.1), ("I", 0.9)])

noise_model = NoiseModel()
noise_model.add_all_qubit_quantum_error(bit_flip, ["h"])

# Build circuit
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

# Run noisy simulation
simulator = AerSimulator(noise_model=noise_model)
compiled_qc = transpile(qc, simulator)

result = simulator.run(
    compiled_qc,
    shots=10000
).result()

print(result.get_counts())
````

# Quantum Teleportation:

````python
from qiskit import *
from qiskit_aer import Aer
from qiskit.visualization import plot_bloch_multivector,array_to_latex
from qiskit.circuit.library import Initialize
from qiskit.quantum_info import random_statevector, Statevector
````

Consider any state $|\psi\rangle$. **Alice** wants to send it to **Bob**. $$|\psi\rangle  = \alpha|0\rangle + \beta|1\rangle $$

````python
# Create separate quantum registers for alice and bob

qr_init = QuantumRegister(1,'q')             # initial qubit to be teleported
qr_alice = QuantumRegister(1, 'alice')       # Alice's qubit (part of the entangled pair)
qr_bob = QuantumRegister(1, 'bob')           # Bob's qubit (part of the entangled pair)

c1 = ClassicalRegister(1,name="c1")          # to store Alice's first measurement result
c2 = ClassicalRegister(1,name="c2")          # to store Alice's second measurement result

qc= QuantumCircuit(qr_init,qr_alice,qr_bob,c1,c2)

display(Math(r"\text{ Quantum Circuit : }"))

display(qc.draw('mpl'))
````

## Step2 -> Now we need to entangle **Alice** and **Bob's** Qubits. For that we are using *hadamard* and *CNOT* gates.

````python
# Step-2
def create_bell_pair(qc,a,b):
    qc.h(a)
    qc.cx(a,b)

create_bell_pair(qc,1,2)

qc.barrier()
display(qc.draw("mpl"))
````

On a Bloch sphere, a qubit in a pure state lies on the surface, while a maximally mixed state is represented by the center. When two qubits are entangled, each qubit individually can be maximally mixed, even though their combined state is pure. The center may not be visibly marked in Qiskit's plot because the Bloch vector has zero length, so no arrow is drawn.


## Step-3 -> Alice performs Reverse EPR operations w.r.t initialzation Qubit



````python
# Step-3
def alice_gates(qc,psi,a):

    qc.cx(psi,a)
    qc.h(psi)
    qc.barrier()

alice_gates(qc,0,1) #qubit 0 control, Qubit 1 target
display(qc.draw("mpl"))
````

## Step-4 -> Alice performs the measurements on her Entangled Qubit and Information carrying Qubit. The results of two measurements are stored in 2 classical bits which will be sent to Bob.

````python
#step-4
def alice_measurements(qc,a,b):
    qc.measure(a, 0)  # Input qubit measurement
    qc.measure(b, 1)  # Alice's entangled qubit measurement

alice_measurements(qc,0,1)
display(qc.draw("mpl"))
````

## Step-5 Bob Gates:

Here,

- if Alice sends $|00\rangle$, Bob should have got the result as $\alpha|0\rangle+\beta|1\rangle$.

- if Alice sends $|01\rangle$, Bob should have got the result as $\alpha|1\rangle+\beta|0\rangle$.

- if Alice sends $|10\rangle$, Bob should have got the result as $\alpha|0\rangle-\beta|1\rangle$.

- if Alice sends $|11\rangle$, Bob should have got the result as $\alpha|1\rangle-\beta|0\rangle$.


Here we also need to apply some unitary gates based on the values Alice sends in order to get exact output.

| Recieved Bits| Result recieved by Bob | Gates to be applied | Result after applying Gates |
|--------------|------------------------|---------------------|-----------------------------|
|00|$\alpha|0\rangle+\beta|1\rangle$|I|$\alpha|0\rangle + \beta|1\rangle$|
|01|$\alpha|1\rangle+\beta|0\rangle$|X|$\alpha|0\rangle + \beta|1\rangle$|
|10|$\alpha|0\rangle-\beta|1\rangle$|Z|$\alpha|0\rangle + \beta|1\rangle$|
|11|$\alpha|1\rangle-\beta|0\rangle$|X&Z|$\alpha|0\rangle + \beta|1\rangle$|

````python
# Step-5

def Bob_gates(qc, qubit, c1, c2):
  with qc.if_test((c2, 1)):
        qc.x(qubit)

  with qc.if_test((c1, 1)):
        qc.z(qubit)


Bob_gates(qc, 2, c1, c2)

display(qc.draw("mpl"))
````
